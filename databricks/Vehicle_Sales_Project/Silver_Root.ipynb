```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
BRONZE_PATH = "abfss://bronze@pravdatalake.dfs.core.windows.net"
```

```python

basic_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/Basic_table"
)

price_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/Price_table"
)

sales_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/Sales_table"
)

ad_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/Ad_table"
)

trim_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/Trim_table"
)
```

```python
print("Basic:", basic_df.count())
print("Price:", price_df.count())
print("Sales:", sales_df.count())
print("Ad:", ad_df.count())
print("Trim:", trim_df.count())

basic_df.printSchema()
```
